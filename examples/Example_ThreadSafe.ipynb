# Threadsafe Decorators

This notebook guides a user on how to use threadsafe decorators when accessing a store in a parallel method.


````python
import kosh
store = kosh.connect("threadsafe.sqlite")
````

## Threadsafe Create

The `**kw_args` are passed into the underlying `kosh.create()` method.

````python
dataset1 = kosh.utils.threadsafe.safe_create(store,
                                             name='my threadsafe dataset',
                                             metadata= {"param1": 100})

dataset2 = kosh.utils.threadsafe.safe_create(store,
                                             name='my threadsafe dataset 2',
                                             metadata= {"param2": 200})
print(dataset1,'\n\n')
print(dataset2)
````

## Threadsafe Associate

The `**kw_args` are passed into the underlying `dataset.associate()` method.

````python
kosh.utils.threadsafe.safe_associate(dataset1,
                                     uri="my_ult_file.ult", 
                                     mime_type="ultra")
print(dataset1)
print(dataset1.list_features())
````

## Threadsafe Get Execution Graph

The `**kw_args` are passed into the underlying `dataset.get_execution_graph()` method.

````python
my_data = kosh.utils.threadsafe.safe_get_execution_graph(dataset1,
                                                         feature="Gaussian (a: 5.0 w: 5.0 c: 0.0)")[:]
print(my_data)
````

## Threadsafe Open

The `**kw_args` are passed into the underlying `store.open()` method.

````python
dataset1_temp = kosh.utils.threadsafe.safe_open(store,
                                                Id=dataset1.id)

print(dataset1.id)
print(dataset1_temp.id)
````

## Threadsafe Create Ensemble

The `**kw_args` are passed into the underlying `store.create_ensemble()` method.

````python
ensemble = kosh.utils.threadsafe.safe_create_ensemble(store, 
                                                      name="My Threadsafe Ensemble", 
                                                      metadata={"root":"/root/path/for/ensemble",
                                                                "project":"Example"})

dataset3 = kosh.utils.threadsafe.safe_create(ensemble,
                                             name='my threadsafe dataset 3',
                                             metadata= {"param1": 300})

dataset4 = kosh.utils.threadsafe.safe_create(ensemble,
                                             name='my threadsafe dataset 4',
                                             metadata= {"param2": 400})

print(ensemble,'\n\n')
print(dataset3,'\n\n')
print(dataset4)
````

## Threadsafe Delete

The `**kw_args` are passed into the underlying `kosh.delete()` method.

````python
kosh.utils.threadsafe.safe_delete(store, dataset1)
````

## Threadsafe Find Datasets

The `**kw_args` are passed into the underlying `ensemble.find_datasets()` method.

````python
ensemble_datasets = kosh.utils.threadsafe.safe_find_datasets(ensemble, name='my threadsafe dataset 4')
print(list(ensemble_datasets)[0])
````

## Threadsafe Find Ensembles

The `**kw_args` are passed into the underlying `store.find_ensembles()` method.

````python
store_ensembles = kosh.utils.threadsafe.safe_find_ensembles(store, name="My Threadsafe Ensemble")
print(list(store_ensembles)[0])
````

## Threadsafe Find

The `**kw_args` are passed into the underlying `kosh.find()` method.

````python
store_find = kosh.utils.threadsafe.safe_find(store, name='my threadsafe dataset 2')
print(list(store_find)[0])
````

## Threadsafe Add

The `**kw_args` are passed into the underlying `kosh.add()` method.

````python
kosh.utils.threadsafe.safe_add(ensemble, dataset2)
print(ensemble)
````

## Threadsafe Clone

The `**kw_args` are passed into the underlying `kosh.clone()` method.

````python
dataset2a = kosh.utils.threadsafe.safe_clone(dataset2)

print(dataset2,'\n\n')
print(dataset2a)
````

## Threadsafe Get

The `*args` and `**kw_args` are passed into the underlying `kosh.get()` method.

````python
dataset2a.add_curve([1,2,3,4], "time", "my_curves")
data = dataset2a.get("time/my_curves")
print(data)
````

## Custom functions with decorators
You can also use the decarators from the above functions for your own custom functions.

### threadsafe_call(num_tries, patience)

```
def threadsafe_call(num_tries, patience):
    """
    A custom thread lock decorator made specifically for Kosh calls.

    Parameters:
    -----------
    num_tries: int
        The number of allowable attempts.
    patience: float
        The number of seconds to wait between tries.

    Returns:
    --------
    A threadsafe version of the decorated function.
    """
```

The decorator `threadsafe_call(num_tries, patience)` below will try to connect to the store 5 times and wait 2 seconds in between each retry to create a dataset using the function `my_custom_create(store, **kw_args)`.

````python
@kosh.utils.threadsafe_decorators.threadsafe_call(5, 2)
def my_custom_create(store, **kw_args):
    return store.create(**kw_args)

dataset_custom = my_custom_create(store,
                                  name='my threadsafe dataset custom',
                                  metadata= {"param1": 300})

print(dataset_custom,'\n\n')
````

### patient_find(num_tries, patience)
```
def patient_find(num_tries, patience):
    """
    Kosh "find" methods can sometimes return empty results when they
    shouldn't. It's really unclear why/how this happens. This decorator
    allows us to perform the search multiple times if the initial results
    are empty.

    Parameters:
    -----------
    num_tries: int
        The number of allowable attempts.
    patience: float
        The number of seconds to wait between tries.

    Returns:
    --------
    A version of the "find" function that will re-attempt
    the find if the initial results are empty.
    """
```

The Kosh find method needs an additional decorator as it can sometimes return empty results. The decorator `threadsafe_call(num_tries, patience)` below will try to connect to the store 5 times and wait 2 seconds in between each retry to find a dataset using the decorator `patient_find(num_tries, patience)` which will try to find a dataset 10 times and wait 3 seconds in between each retry to find a dataset using the function `my_custom_find(store, **kw_args)`.

````python
@kosh.utils.threadsafe_decorators.patient_find(10, 3)
@kosh.utils.threadsafe_decorators.threadsafe_call(5, 2)
def my_custom_find(store, **kw_args):
    return iter(list(store.find(**kw_args)))

store_find = my_custom_find(store, name='my threadsafe dataset custom')
print(list(store_find)[0])
````
